```python
X=dataset.iloc[:,:-1]
Y=dataset.iloc[:,-1]
```

```python
X.head(

)
```

train test split

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train, Y_test=train_test_split(X,Y,test_size=0.3,random_state=42)
```

standarizing

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
import pickle
pickle.dump(scaler,open('scaling.pkl','wb'))
```

model train

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
regression=LinearRegression()
regression.fit(X_train,Y_train)
```

```python
print(regression.coef_)
```

```python
print(regression.intercept_)
```

```python
## on which parameter model has been train
regression.get_params()
```

```python
##prediction
reg_prdict=regression.predict(X_test)
```

```python
reg_prdict
```

```python
## plot scatter plot for pridiction
plt.scatter(Y_test,reg_prdict)
```

```python
##residual
residuals=Y_test-reg_prdict
```

```python
residuals
```

```python
##plot residual
sns.displot(residuals,kind='kde')
```

```python
## scater plot with respect to prediction and residual
## uniform distribution
plt.scatter(reg_prdict,residuals)
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import root_mean_squared_error
print(mean_absolute_error(Y_test,reg_prdict))
print(mean_squared_error(Y_test,reg_prdict))
print(root_mean_squared_error(Y_test,reg_prdict))
```

```python
## R squared and adjusted R square
from sklearn.metrics import r2_score
score=r2_score(Y_test,reg_prdict)
print(score)
```

adjusted r2

```python
1-(1-score)*(len(Y_test)-1)/(len(Y_test)-X_test.shape[1]-1)
```
